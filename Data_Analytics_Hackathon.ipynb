```python
import pandas as pd

swap_file = dataset_path + "/swap_events.csv"

total_events = 0
event_counts = {}
tariff_counts = {}
sync_counts = {}
missing_counts = None

for chunk in pd.read_csv(swap_file, chunksize=100000):

    total_events += len(chunk)

    # Event types
    counts = chunk["event_type"].value_counts()
    for key, value in counts.items():
        event_counts[key] = event_counts.get(key, 0) + value

    # Tariff types
    counts = chunk["tariff_code"].value_counts()
    for key, value in counts.items():
        tariff_counts[key] = tariff_counts.get(key, 0) + value

    # Sync modes
    counts = chunk["sync_mode"].value_counts()
    for key, value in counts.items():
        sync_counts[key] = sync_counts.get(key, 0) + value

    # Missing values
    current_missing = chunk.isna().sum()

    if missing_counts is None:
        missing_counts = current_missing
    else:
        missing_counts = missing_counts.add(current_missing, fill_value=0)

print("Total events:", total_events)

print("\nEvent types:")
print(pd.Series(event_counts).sort_values(ascending=False))

print("\nTariff codes:")
print(pd.Series(tariff_counts).sort_values(ascending=False))

print("\nSync modes:")
print(pd.Series(sync_counts).sort_values(ascending=False))

print("\nMissing values:")
print(missing_counts.astype(int))
```

```python
station_counts = {}

for chunk in pd.read_csv(
    swap_file,
    usecols=["station_id"],
    chunksize=100000
):
    counts = chunk["station_id"].value_counts()

    for station, count in counts.items():
        station_counts[station] = station_counts.get(station, 0) + count

station_counts = pd.Series(station_counts).sort_values(ascending=False)

print("Number of stations:", len(station_counts))

print("\nTop 15 busiest stations:")
print(station_counts.head(15))
```

```python
stations = pd.read_csv(
    dataset_path + "/stations.csv"
)

print("Stations dataset shape:", stations.shape)

print("\nColumns:")
print(stations.columns.tolist())

display(stations.head())
```

```python
# Convert station counts into a proper DataFrame

station_counts_df = station_counts.reset_index()

station_counts_df.columns = ["station_id", "total_events"]

# Merge station information with total swap events

station_analysis = stations.merge(
    station_counts_df,
    on="station_id",
    how="left"
)

# If any station has no recorded events, set it to 0

station_analysis["total_events"] = station_analysis["total_events"].fillna(0)

# Display basic information

print("Rows:", len(station_analysis))
print("Columns:", len(station_analysis.columns))

# Display the 15 busiest stations

display(
    station_analysis[
        [
            "station_id",
            "city",
            "zone",
            "location_type",
            "host_type",
            "slots_2w",
            "slots_3w",
            "inventory_target_2w",
            "inventory_target_3w",
            "monthly_rent_inr",
            "monthly_maintenance_inr",
            "grid_tariff_inr_kwh",
            "connectivity_tier",
            "total_events"
        ]
    ]
    .sort_values("total_events", ascending=False)
    .head(15)
)
```

```python
station_analysis["total_slots"] = (
    station_analysis["slots_2w"] +
    station_analysis["slots_3w"]
)

station_analysis["events_per_slot"] = (
    station_analysis["total_events"] /
    station_analysis["total_slots"]
)

display(
    station_analysis[
        [
            "station_id",
            "city",
            "zone",
            "total_events",
            "slots_2w",
            "slots_3w",
            "total_slots",
            "events_per_slot"
        ]
    ]
    .sort_values("events_per_slot", ascending=False)
    .head(15)
)
```

```python
station_outcomes = {}

for chunk in pd.read_csv(
    swap_file,
    usecols=["station_id", "event_type"],
    chunksize=100000
):
    grouped = (
        chunk
        .assign(
            unsuccessful=chunk["event_type"] != "swap_completed"
        )
        .groupby("station_id")
        .agg(
            total_events=("event_type", "size"),
            unsuccessful_events=("unsuccessful", "sum")
        )
    )

    for station, row in grouped.iterrows():

        if station not in station_outcomes:
            station_outcomes[station] = {
                "total_events": 0,
                "unsuccessful_events": 0
            }

        station_outcomes[station]["total_events"] += int(row["total_events"])
        station_outcomes[station]["unsuccessful_events"] += int(
            row["unsuccessful_events"]
        )

station_outcomes_df = pd.DataFrame.from_dict(
    station_outcomes,
    orient="index"
)

station_outcomes_df.index.name = "station_id"

station_outcomes_df["unsuccessful_rate_pct"] = (
    station_outcomes_df["unsuccessful_events"]
    / station_outcomes_df["total_events"]
    * 100
)

station_outcomes_df = station_outcomes_df.reset_index()

print("Stations analyzed:", len(station_outcomes_df))

display(
    station_outcomes_df
    .sort_values("unsuccessful_rate_pct", ascending=False)
    .head(15)
)
```

```python
city_analysis = station_outcomes_df.merge(
    stations[["station_id", "city"]],
    on="station_id",
    how="left"
)

city_analysis = (
    city_analysis
    .groupby("city")
    .agg(
        total_events=("total_events", "sum"),
        unsuccessful_events=("unsuccessful_events", "sum")
    )
    .reset_index()
)

city_analysis["unsuccessful_rate_pct"] = (
    city_analysis["unsuccessful_events"]
    / city_analysis["total_events"]
    * 100
)

city_analysis["completed_events"] = (
    city_analysis["total_events"]
    - city_analysis["unsuccessful_events"]
)

display(
    city_analysis
    .sort_values("unsuccessful_rate_pct", ascending=False)
)
```

```python
# Create station -> city mapping
station_city = stations.set_index("station_id")["city"]

city_event_counts = {}

for chunk in pd.read_csv(
    swap_file,
    usecols=["station_id", "event_type"],
    chunksize=100000
):
    chunk["city"] = chunk["station_id"].map(station_city)

    grouped = (
        chunk
        .groupby(["city", "event_type"])
        .size()
    )

    for (city, event_type), count in grouped.items():
        key = (city, event_type)
        city_event_counts[key] = (
            city_event_counts.get(key, 0) + count
        )

city_event_df = (
    pd.Series(city_event_counts, name="event_count")
    .reset_index()
)

city_event_df.columns = [
    "city",
    "event_type",
    "event_count"
]

# Calculate percentage within each city
city_event_df["city_total"] = (
    city_event_df.groupby("city")["event_count"]
    .transform("sum")
)

city_event_df["percentage"] = (
    city_event_df["event_count"]
    / city_event_df["city_total"]
    * 100
)

display(
    city_event_df
    .sort_values(
        ["city", "event_count"],
        ascending=[True, False]
    )
)
```

```python
city_station_summary = (
    station_analysis
    .groupby("city")
    .agg(
        total_events=("total_events", "sum"),
        total_slots=("total_slots", "sum"),
        number_of_stations=("station_id", "count")
    )
    .reset_index()
)

city_station_summary["events_per_slot"] = (
    city_station_summary["total_events"]
    / city_station_summary["total_slots"]
)

city_station_summary = city_station_summary.merge(
    city_analysis[
        ["city", "unsuccessful_rate_pct"]
    ],
    on="city",
    how="left"
)

display(
    city_station_summary
    .sort_values("unsuccessful_rate_pct", ascending=False)
)
```

```python
hourly_demand = {}

for chunk in pd.read_csv(
    swap_file,
    usecols=["station_id", "event_ts", "event_type"],
    chunksize=100000
):
    chunk["hour_start"] = pd.to_datetime(chunk["event_ts"]).dt.floor("h")

    grouped = (
        chunk
        .groupby(["station_id", "hour_start"])
        .agg(
            total_events=("event_type", "size"),
            unsuccessful_events=(
                "event_type",
                lambda x: (x != "swap_completed").sum()
            )
        )
        .reset_index()
    )

    for row in grouped.itertuples(index=False):
        key = (row.station_id, row.hour_start)

        if key not in hourly_demand:
            hourly_demand[key] = [0, 0]

        hourly_demand[key][0] += row.total_events
        hourly_demand[key][1] += row.unsuccessful_events

hourly_demand_df = pd.DataFrame(
    [
        [station, hour, values[0], values[1]]
        for (station, hour), values in hourly_demand.items()
    ],
    columns=[
        "station_id",
        "hour_start",
        "total_events",
        "unsuccessful_events"
    ]
)

hourly_demand_df["unsuccessful_rate_pct"] = (
    hourly_demand_df["unsuccessful_events"]
    / hourly_demand_df["total_events"]
    * 100
)

print("Rows created:", len(hourly_demand_df))

display(hourly_demand_df.head())
```

```python
status_file = dataset_path + "/station_hourly_status (1).csv"

status_cols = [
    "station_id",
    "hour_start",
    "charged_2w_avg",
    "charged_2w_min",
    "charged_3w_min",
    "packs_charging",
    "packs_quarantined",
    "chargers_online",
    "ambient_temp_c",
    "cabinet_temp_c",
    "avg_charge_minutes",
    "outage_minutes",
    "telemetry_status"
]

station_status = pd.read_csv(
    status_file,
    usecols=status_cols
)

station_status["hour_start"] = pd.to_datetime(
    station_status["hour_start"]
)

print("Station-hour rows:", len(station_status))

display(station_status.head())
```

```python
hourly_analysis = hourly_demand_df.merge(
    station_status,
    on=["station_id", "hour_start"],
    how="left"
)

print("Rows after merge:", len(hourly_analysis))

print("\nMissing status values:")
display(
    hourly_analysis[
        [
            "charged_2w_avg",
            "charged_2w_min",
            "charged_3w_min",
            "packs_charging",
            "chargers_online",
            "outage_minutes"
        ]
    ].isna().sum()
)

display(hourly_analysis.head())
```

```python
availability_analysis = hourly_analysis.dropna(
    subset=["charged_2w_avg"]
).copy()

availability_analysis["battery_availability"] = pd.qcut(
    availability_analysis["charged_2w_avg"],
    q=3,
    labels=["Low", "Medium", "High"]
)

availability_summary = (
    availability_analysis
    .groupby("battery_availability", observed=True)
    .agg(
        station_hours=("station_id", "count"),
        total_events=("total_events", "sum"),
        unsuccessful_events=("unsuccessful_events", "sum")
    )
    .reset_index()
)

availability_summary["unsuccessful_rate_pct"] = (
    availability_summary["unsuccessful_events"]
    / availability_summary["total_events"]
    * 100
)

display(availability_summary)
```

```python
failed_battery_hourly = {}

for chunk in pd.read_csv(
    swap_file,
    usecols=["station_id", "event_ts", "event_type"],
    chunksize=100000
):
    chunk["hour_start"] = pd.to_datetime(
        chunk["event_ts"]
    ).dt.floor("h")

    failed = chunk[
        chunk["event_type"] == "failed_no_charged_battery"
    ]

    grouped = (
        failed
        .groupby(["station_id", "hour_start"])
        .size()
        .reset_index(name="failed_no_charged_battery")
    )

    for row in grouped.itertuples(index=False):
        key = (row.station_id, row.hour_start)

        if key not in failed_battery_hourly:
            failed_battery_hourly[key] = 0

        failed_battery_hourly[key] += row.failed_no_charged_battery

failed_battery_df = pd.DataFrame(
    [
        [station, hour, count]
        for (station, hour), count in failed_battery_hourly.items()
    ],
    columns=[
        "station_id",
        "hour_start",
        "failed_no_charged_battery"
    ]
)

print("Station-hours with failed battery events:", len(failed_battery_df))

display(failed_battery_df.head())
```

```python
failed_battery_analysis = hourly_analysis[
    [
        "station_id",
        "hour_start",
        "total_events"
    ]
].merge(
    failed_battery_df,
    on=["station_id", "hour_start"],
    how="left"
)

failed_battery_analysis["failed_no_charged_battery"] = (
    failed_battery_analysis["failed_no_charged_battery"]
    .fillna(0)
)

failed_battery_analysis = failed_battery_analysis.merge(
    availability_analysis[
        [
            "station_id",
            "hour_start",
            "charged_2w_avg",
            "battery_availability"
        ]
    ],
    on=["station_id", "hour_start"],
    how="left"
)

failed_battery_summary = (
    failed_battery_analysis
    .groupby("battery_availability", observed=True)
    .agg(
        station_hours=("station_id", "count"),
        total_events=("total_events", "sum"),
        failed_no_charged_battery=(
            "failed_no_charged_battery",
            "sum"
        )
    )
    .reset_index()
)

failed_battery_summary["failure_rate_pct"] = (
    failed_battery_summary["failed_no_charged_battery"]
    / failed_battery_summary["total_events"]
    * 100
)

display(failed_battery_summary)
```

### 3. Temperature-Aware Operations
Increase monitoring and operational preparedness during high-temperature periods.

Track battery SOH and plan replacement or retirement as batteries approach the observed retirement range.

Use `no_battery_available` support tickets to identify stations where battery availability may require intervention.
